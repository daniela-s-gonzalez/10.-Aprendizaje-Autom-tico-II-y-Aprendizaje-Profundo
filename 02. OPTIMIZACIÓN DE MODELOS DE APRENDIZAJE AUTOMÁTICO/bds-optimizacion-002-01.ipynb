 ## **<font color="DeepPink">Definición de overfitting, generalizacion y underfitting</font>**

<p align="justify">
En <code>Scikit-learn</code> y para el aprendizaje automático, los términos <b>overfitting</b> (sobreajuste), <b>generalización</b> y <b>underfitting</b> (subajuste) se refieren a fenómenos relacionados con el rendimiento de un modelo.
<br><br>
✅ El <b>overfitting</b> ocurre cuando un modelo se ajusta demasiado a los datos de entrenamiento y se vuelve muy específico para esos datos, perdiendo la capacidad de generalizar correctamente a los datos nuevos, es decir los datos no vistos. En otras palabras, el modelo "aprende de memoria" los datos de entrenamiento en lugar de capturar los patrones subyacentes. Como resultado, el modelo puede tener un rendimiento muy alto en los datos de entrenamiento, pero un rendimiento deficiente en los datos no vistos, lo que reduce su capacidad de generalización. En <code>Scikit-learn</code>, el sobreajuste se puede identificar cuando el rendimiento del modelo en el conjunto de entrenamiento es significativamente mejor que en el conjunto de evaluación.
<br><br>
✅ La <b>generalización</b> se refiere a la capacidad de un modelo para realizar predicciones precisas en datos no vistos, es decir, su capacidad para capturar los patrones subyacentes y aplicarlos a nuevas instancias. Un modelo bien generalizado es capaz de adaptarse a nuevos datos y realizar predicciones precisas sin haberlos visto previamente. En <code>Scikit-learn</code>, la generalización se evalúa mediante el rendimiento del modelo en un conjunto de evaluación independiente.
<br><br>
✅ El <b>underfitting</b> ocurre cuando un modelo es demasiado simple para capturar los patrones complejos presentes en los datos de entrenamiento. En otras palabras, el modelo no se ajusta lo suficiente a los datos de entrenamiento y no logra capturar las relaciones importantes. Como resultado, el modelo puede tener un rendimiento deficiente tanto en los datos de entrenamiento como en los datos de evaluación. En <code>Scikit-learn</code>, el subajuste se puede identificar cuando el rendimiento del modelo es bajo tanto en el conjunto de entrenamiento como en el conjunto de evaluación.
<br><br>

👀 El objetivo deseado en el aprendizaje automático es lograr un equilibrio entre overfitting y underfitting, es decir, obtener un modelo que generalice bien a nuevos datos sin ajustarse demasiado a los datos de entrenamiento. Esto se logra mediante técnicas como la selección adecuada de variables explicativas, la validación cruzada y la búsqueda de hiper-parámetros. <code>Scikit-learn</code> proporciona herramientas y métodos para ayudar a abordar estos desafíos y encontrar el equilibrio óptimo entre el sobreajuste y el subajuste.

 # **<font color="DeepPink">Overfitting versus Underfitting</font>**


<p align="justify">
💡 Para comprender mejor el rendimiento de generalización de nuestro modelo y tal vez encontrar información sobre cómo mejorarlo, comparamos el error de prueba con el error de entrenamiento. Por lo tanto, necesitamos calcular el error en el conjunto de entrenamiento, lo cual es posible usando la función <code>cross_validate</code>.

 ## **<font color="DeepPink">Creación y ajuste del modelo</font>**

Para resolver el problema de regresión, utilizaremos un [Árbol de Regresión](https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeRegressor.html)

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model = DecisionTreeRegressor(random_state=0)
model
```

```python
from sklearn.model_selection import cross_validate, ShuffleSplit
```

```python
cv = ShuffleSplit(n_splits=30, test_size=0.2)

cv_results = cross_validate(model, X, y,
                            cv=cv,
                            scoring="neg_mean_absolute_error",
                            return_train_score=True,
                            n_jobs=2)

cv_results = pd.DataFrame(cv_results)
```

<p align="justify">
El hiperparámetro <code>return_train_score</code> en la función <code>cross_validate</code> determina si se deben devolver o no los puntajes de entrenamiento en los resultados de validación cruzada. Los puntajes de entrenamiento representan el rendimiento del modelo en los datos utilizados para entrenarlo en la validación cruzada.

Si <code>return_train_score=True</code>, los puntajes de entrenamiento se incluirán en el diccionario de resultados devuelto por la función <code>cross_validate</code>. Esto puede ser útil si deseas examinar los puntajes de entrenamiento junto con los puntajes de prueba para evaluar el rendimiento y la capacidad de generalización del modelo.

El parámetro <code>n_jobs</code> en <code>cross_validate</code> especifica el número de núcleos de CPU que se utilizarán para paralelizar el proceso de validación cruzada. Al establecerlo en 2, estás indicando a <code>sklearn</code> que utilice dos núcleos de CPU para la ejecución paralela. Esto puede ayudar a acelerar el proceso de validación cruzada, especialmente para modelos con una alta carga computacional o conjuntos de datos grandes.

```python
cv_results.head()
```

<p align="justify">
👀 La validación cruzada utilizó el error absoluto medio negativo. Transformamos el error absoluto medio negativo en un error absoluto medio positivo.

```python
scores = pd.DataFrame()
scores[["train error", "test error"]] = -cv_results[["train_score", "test_score"]]
```

```python
scores.head()
```

 ## **<font color="DeepPink">Histograma Overfitting versus Underfitting</font>**

```python
fig = make_subplots(rows=1, cols=2,
                    subplot_titles=[f"{i}" for i in scores.columns],
                    x_title="MAE $")
```

```python
for i,j in enumerate(scores.columns):
  fig.add_trace(go.Histogram(x=scores[j],
                             name=j),
                row=1, col=i+1)
```

```python
fig.update_layout(template="gridon",
                  title_text="Overfitting versus Underfitting",
                  bargap=0.2)

fig.show()
```

<p align="justify">
👀 Al graficar la distribución de los errores de entrenamiento y los errores de prueba, se puede obtener información sobre si nuestro modelo se ajusta demasiado (Overfitting), se ajusta poco (Underfitting) o los dos al mismo tiempo.
<br><br>
✅ Aquí, observamos un <b>pequeño error de entrenamiento</b> (en realidad cero), lo que significa que el modelo <b>no tiene underfitting</b>. El modelo es lo suficientemente flexible para capturar cualquier variación presente en el conjunto de entrenamiento.
<br><br>
✅ Sin embargo, el <b>error de prueba</b> <mark>es significativamente mayor, lo que nos dice que el modelo tiene <b>overfitting</b></mark>. El modelo ha memorizado muchas variaciones del conjunto de entrenamiento que podría considerarse "ruido" porque no generaliza en una buena predicción en el conjunto de prueba.

 ## **<font color="DeepPink">Curva de validación</font>**

<p align="justify">
La <b>curva de validación</b> en <code>Scikit-learn</code> es una herramienta utilizada para evaluar el rendimiento de un modelo de aprendizaje automático en función de diferentes valores de un hiperparámetro específico. Proporciona una visualización gráfica de cómo varía la puntuación de rendimiento del modelo en el conjunto de entrenamiento y el conjunto de prueba a medida que se modifican los valores del hiperparámetro.
<br><br>
<code>Scikit-learn</code> ofrece la función <code>validation_curve</code> del módulo <code>model_selection</code> para generar la curva de validación. Esta función permite realizar un análisis sistemático del rendimiento del modelo en función de un hiperparámetro, como la profundidad máxima de un árbol de decisión o la cantidad de vecinos en un algoritmo de $k$ vecinos más cercanos (KNN por sus siglas en inglés).

<p align="justify">
Algunos hiperparámetros del modelo suelen ser la clave para pasar de un modelo que subajuste a un modelo que sobreajuste, pasando por una región donde se puede conseguir un buen equilibrio entre los dos. Se puede adquirir una idea trazando una curva llamada la <b>curva de validación</b>.

```python
from sklearn.model_selection import validation_curve
```

<p align="justify">
🛑 En un árbol de decisión, el parámetro <code>max_depth</code> se utiliza para controlar el equilibrio entre sobreajuste y subajuste del modelo.

```python
max_depth = [1, 5, 10, 15, 20, 25] # es la complejidad del modelo árbol
```

```python
train_scores, test_scores = validation_curve(model,
                                             X, y,
                                             param_name="max_depth",
                                             param_range=max_depth,
                                             cv=cv,
                                             scoring="neg_mean_absolute_error",
                                             n_jobs=2)
```

```python
train_errors, test_errors = -train_scores, -test_scores
```

<p align="justify">
📊 Ahora que recopilamos los resultados, mostramos la curva de validación al graficar los errores de entrenamiento y los errores de prueba (así como sus desviaciones).

```python
fig = go.Figure()

fig.add_trace(go.Scatter(x=max_depth,
                         y=train_errors.mean(axis=1),
                         name='train_errors'))

fig.add_trace(go.Scatter(x=max_depth,
                         y=test_errors.mean(axis=1),
                         name='test_errors'))

fig.update_layout(template="gridon",
                  title_text="Validation curve for Decision Tree",
                  xaxis_title='Maximum depth of Decision Tree',
                  yaxis_title='Mean Absolute Error (k$)'
                  )

fig.show()
```

👀 otra forma de graficar

```python
tupla = (("train_errors", train_errors),("test_errors",test_errors))
```

```python
fig = go.Figure()

for i, j in tupla:
  fig.add_trace(go.Scatter(x=max_depth,
                           y=j.mean(axis=1),
                           name=f"{i}"))

fig.update_layout(template="gridon",
                  title_text="Validation curve for Decision Tree",
                  xaxis_title='Maximum depth of Decision Tree',
                  yaxis_title='Mean Absolute Error (k$)'
                  )

fig.show()
```

<p align="justify">
👀 La curva de validación se puede dividir en tres áreas:
<br><br>
🥇 Para <code>max_depth < 10</code>, el árbol de decisión se subajusta. El error de entrenamiento y prueba son altos. El modelo tiene poca profundidad, por lo que no puede capturar gran parte de la variabilidad de la variable objetivo.
<br><br>
🥈 La región alrededor de <code>max_ depth = 10</code> corresponde al parámetro para el cual el árbol generaliza mejor. En este punto, el modelo posee la flexibilidad suficiente para capturar una fracción significativa de la variabilidad de la variable objetivo, evitando memorizar el ruido presente en los datos.
<br><br>
🥉 Para <code>max_depth > 10</code>, el árbol de decisión se sobreajusta. El error de entrenamiento se vuelve muy pequeño, mientras que el error de prueba aumenta. En esta región, el modelo crea decisiones específicas para las muestras ruidosas, lo cual afecta su capacidad de generalización en los datos de prueba.



<p align = "justify">
👀 Tenga en cuenta que para <code>max_ depth = 10</code>, el modelo se sobreajusta un poco ya que hay una brecha entre el error de entrenamiento y el error de prueba.
<br><br>
<mark>Este modelo puede llegar a ser deficiente, porque el error de entrenamiento todavía está lejos de cero (más de 30k $),</mark> lo cual significa que el modelo todavía podría estar demasiado limitado para modelar partes interesantes de los datos. Sin embargo, el error de prueba es el menor. Esto es lo mejor que podíamos lograr simplemente ajustando el hiperparámetro <code>max_ depth</code>.
<br><br>
Por otro lado, mirar las medias de los errores es bastante limitado. También deberíamos observar la desviación estándar para evaluar la dispersión de la puntuación.

```python
fig = make_subplots(rows=2, cols=1,
                    x_title="Maximum depth of Decision Tree",
                    y_title='Mean Absolute Error (k$)')
```

```python
fig.add_trace(go.Scatter(x=max_depth,
                         y=train_errors.mean(axis=1),
                         name='train_errors mean'),
              row=1, col=1)

fig.add_trace(go.Scatter(x=max_depth,
                         y=test_errors.mean(axis=1),
                         name='test_errors mean'),
              row=1, col=1)

fig.add_trace(go.Scatter(x=max_depth,
                         y=train_errors.std(axis=1),
                         name='train_errors std'),
              row=2, col=1)

fig.add_trace(go.Scatter(x=max_depth,
                         y=test_errors.std(axis=1),
                         name='test_errors std'),
              row=2, col=1)


fig.update_layout(template="gridon",
                  title_text="Validation curve for Decision Tree")

fig.show()
```

<p align="justify">
✅ Las desviaciones estándar reflejan la dispersión de los errores para cada valor del hiperparámetro evaluado. Proporcionan una medida de la variabilidad del rendimiento del modelo. Cuanto mayor sea la desviación estándar, mayor será la dispersión de los errores y, por lo tanto, mayor la incertidumbre asociada al rendimiento del modelo.
<br><br>
🛑 En este ejemplo, la desviación estándar de los errores es pequeña en comparación con sus respectivos valores y, por lo tanto, las conclusiones anteriores son bastante claras.
<br><br>
La curva de validación resultante se puede visualizar para analizar el comportamiento del rendimiento en función de los valores del hiperparámetro y, a partir de ello, tomar decisiones informadas sobre la configuración óptima del modelo.
