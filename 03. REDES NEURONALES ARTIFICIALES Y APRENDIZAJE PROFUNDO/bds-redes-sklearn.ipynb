 # **<font color="DeepPink">Redes Neuronales con Scikit-learn</font>**

<p align="justify">
En el ecosistema de python, existen múltiples bibliotecas que permiten crear modelos basados en redes neuronales. Conviene diferenciar entre dos casos de uso ya que, dependiendo de estos, son más adecuadas unas bibliotecas u otras:
<ul p align="justify">
<li><b>Modelos de redes simples</b>: estos modelos se caracterizan por tener arquitecturas relativamente sencillas por lo que los requerimientos computacionales no son elevados y no es necesario el uso de GPUs. Dentro de este grupo destaca <b>Scikit-learn</b>.

<li><b>Deep learning</b>: son modelos más complejos (redes convolucionales, redes recurrentes, LSTM, etc.) cuyos requerimientos computacionales hacen necesario el uso de GPUs. Para este tipo de modelos se tiene que recurrir a frameworks especializados para como <b>Keras</b>, <b>Tensorflow</b> o <b>Pytorch</b>.</ul>
<p align="justify">
Los ejemplos de este colab se corresponden con el primer caso de uso, por lo que se utiliza la biblioteca Scikit-learn para crear los modelos.

 ## **<font color="DeepPink">Hiperparámetros más importantes de las redes neuronales</font>**

<p align="justify">
La gran "flexibilidad" que tienen las redes neuronales es un arma de doble filo. Por un lado, son capaces de generar modelos que aprenden relaciones muy complejas, sin embargo, sufren fácilmente el problema de <b>sobreajuste (overfitting)</b> lo que los incapacita al tratar de predecir nuevas observaciones. La forma de minimizar este problema y conseguir modelos útiles, es configurar de forma adecuada sus hiperparámetros. Algunos de los más importantes son:
<ul p align="justify">
<li><b>Número y tamaño de capas</b>: la arquitectura de una red, el número de capas y el número de neuronas que forman parte de cada capa, determinan en gran medida la complejidad del modelo y con ello su potencial capacidad de aprendizaje. La capa de entrada tiene tantas neuronas como variables predictoras y la capa de salida tiene una neurona en problemas de regresión y tantas como clases en problemas de clasificación. El usuario suele especificar únicamente el número de capas ocultas y el tamaño de las mismas (cantidad de neuronas).<br>
Cuantas más neuronas y capas, mayor la complejidad de las relaciones que puede aprender el modelo. Sin embargo, dado que en cada neurona está conectada al resto de neuronas de las capas adyacentes, el número de pesos a aprender aumenta y con ello el tiempo de entrenamiento.
<li><b>Learning rate</b>: o tasa de aprendizaje establece cómo de rápido pueden cambiar los pesos de un modelo a medida que se optimiza (aprende). Este hiperparámetro es uno de los más complicados de establecer, ya que depende mucho de los datos e interacciona con el resto de hiperparámetros. Si el learning rate es muy grande, el proceso de optimización puede ir saltando de una región a otra sin que el modelo sea capaz de aprender. Si por el contrario, el learning rate es muy pequeño, el proceso de entrenamiento puede tardar demasiado y no llegar a completarse.
<li><b>Algoritmo de optimización</b>: el descenso de gradiente fue uno de los primeros métodos de optimización utilizados para entrenar modelos de redes neuronales. Utiliza directamente el gradiente para dirigir la optimización, pronto se vio que esto genera problemas a medida que las redes aumentan de tamaño (neuronas y capas). En muchas regiones del espacio de búsqueda, el gradiente es muy proximo a cero, lo que hace que la optimización quede estancada en estas regiones. Para evitar este problema, se han desarrollado modificaciones del descenso de gradiente capaces de adaptar el learning rate en función del gradiente y subgradiente. De esta forma, el proceso de aprendizaje se ralentiza o acelera dependiendo de las características de la región del espacio de búsqueda en el que se encuentren. Aunque existen multitud de adaptaciones, suele recomendarse para conjuntos de datos pequeños 'lbfgs' y para conjuntos de datos grandes 'adam'.<br>
Puede encontrar una descripción bien detallada de los algoritmos de optimización en el libro <a href="https://d2l.ai/chapter_optimization/index.html">Dive into Deep Learning</a>
<li><b>Regularización</b>: los métodos de regularización se utilizan con el objetivo de reducir el sobreajuste (overfitting) de los modelos. Un modelo con sobreajuste memoriza los datos de entrenamiento pero es incapaz de predecir correctamente nuevas observaciones. En los modelos de redes neuronales las estrategias de regularización son fundamentales. De entre las muchas que existen, destacan la regularización L1 y L2 (weight decay), y el dropout.

* **Penalización L1 y L2**: el objetivo de la penalización L1 y L2, esta última también conocida como weight decay, es evitar que los pesos tomen valores excesivamente elevados. De esta forma se evita que unas pocas neuronas dominen el comportamiento de la red y se fuerza a que las variables predictoras poco relevantes (ruido) tengan pesos próximos o iguales a cero.

* **Dropout**: el proceso consiste en de desactivar aleatoriamente una serie de neuronas durante el proceso de entrenamiento. En concreto, durante cada iteración del entrenamiento, se ponen a cero los pesos de una fracción aleatoria de neuronas por capa. El método de dropout (Srivastava et al. 2014) se ha convertido en un estándar para entrenar redes neuronales. El porcentaje de neuronas que suele desactivarse por capa (dropout rate) suele ser un valor entre 0.2 y 0.5.

 ## **<font color="DeepPink">Modelos de redes neuronales en Scikit-learn</font>**

<p align="justify">
Para crear modelos basado en redes neuronales con scikit-learn, se utilizan <a href="https://scikit-learn.org/stable/modules/generated/sklearn.neural_network.MLPRegressor.html">sklearn.neural_network.MLPRegressor</a> para regresión y <a href="https://scikit-learn.org/stable/modules/generated/sklearn.neural_network.MLPClassifier.html">sklearn.neural_network.MLPClassifier</a> para clasificación.
<br><br>
Son muchos los hiperparámetros que controlan el comportamiento de este tipo de modelos. Afortunadamente, los responsables de su implementación han establecido valores por defecto que suelen funcionar adecuadamente en muchos escenarios. A continuación, se muestran los más influyentes:
<ul p align="justify">
<li><code>hidden_layer_sizes</code>: número y tamaño de las capas ocultas. Por ejemplo, (100) para una única capa oculta con 100 neuronas, y (16, 16) para dos capas ocultas de 16 neuronas de cada una. El valor por defecto es (100,).
<li><code>activation</code>: función de activación de las capas ocultas. Puede ser:

*   'identity': sin activación ($f(x) = x$),
*   'logistic': función sigmoide ($f(x) = 1 / (1 + e^{-x})$),
*   'tanh': función tangente hiperbólica ($f(x) = tanh(x)$),
*   'relu': función Rectified linear unit ($f(x) = max(0, x)$).
<br></ul>
<p align="justify">
La misma función de activación se aplica a todas las capas ocultas, no se permiten distintas. El valor por defecto es 'relu'.

<ul p align="justify">
<li><code>solver</code>: el algoritmo de optimización utilizado para aprender los pesos y bias de la red. Puede ser:

*  'lbfgs': de la familia de los métodos de Newton,
*  'sgd':  gradiente descendente estocástico,
*  'adam': gradiente descendente estocástico propuesto por Kingma, Diederik y Jimmy Ba.

Por defecto se utiliza 'adam', que es el que mejores resultados suele dar para conjuntos de datos con miles de observaciones. Para sets de datos pequeños, 'lbfgs' converge más rápido y puede conseguir mejores resultados.
</ul>
<ul p align="justify">
<li><code>alpha</code>: regularización L2 (weight decay). El valor por defecto es 0.0001.
<li><code>batch_size</code>: tamaño de batch utilizado en los solvers estocásticos ('sgd' y 'adam'). Este parámetro se ignora si el solver es 'lbfgs'. Por defecto se utiliza min(200, n_samples).
<li><code>learning_rate</code>: estrategia para modificar el learning rate durante el entrenamiento. Solo se utiliza cuando solver='sgd'. Puede ser:

* 'constant': se utiliza el valor especificado en el hiperparámetro <code>learning_rate_init</code> durante todo el proceso de entrenamiento.

* 'invscaling': se reduce progresivamente el learning rate en cada iteración.

* 'adaptive': mantiene constante el valor especificado en el hiperparámetro <code>learning_rate_init</code> siempre y cuando el modelo siga mejorando (reducción de la función de coste). Si entre dos epochs consecutivas el modelo no mejora un mínimo definido en el argumento tol, el learning rate se divide por 5.

</ul>
<ul p align="justify">
<li><code>learning_rate_init</code>: valor inicial de learning rate. Solo se utiliza cuando el solver es 'sgd' o 'adam'. Por defecto el valor es 0.001.
<li><code>power_t</code>: exponente eutilizado para reducir el learning rate cuando <code>learning_rate='invscaling'</code>. Por defecto se emplea el valor 0.5. Solo se utiliza este argumento cuando solver='sgd'.
<li><code>max_iter</code>: número máximo de iteraciones de entrenamiento. Para los solvers estocásticos ('sgd' y 'adam') este valor se corresponde con el número de epochs (cuantas veces participa en el entrenamiento cada observación). Se emplean por defecto 200.
<li><code>shuffle</code>: si se mezclan aleatoriamente las observaciones en cada iteración. Por defecto es True.
<li><code>random_state</code>: semilla utilizada para todos los pasos del entrenaiento que requieren de valores aleatorios (inicialización de pesos y sesgos).
<li><code>tolfloat</code>: valor de tolerancia utilizado en la optimización. Si la función de coste no mejora durante <code>n_iter_no_change</code> iteraciones consecutivas un mínimo de <code>tolfloat</code>, el entrenamiento finaliza. Por defecto se emplea 0.0001.
<li><code>early_stopping</code>: detener el entrenamiento cuando la métrica de validación no mejora. Automáticamente, se separa un porcentaje de <code>validation_fraction</code> del conjunto de entrenamiento que se utiliza como conjunto de validación. Si durante más de <code>n_iter_no_change</code> iteraciones (epochs), la métrica de validación no mejora un mínimo de <code>tol</code>, el entrenamiento termina. Solo se aplica si solver es 'sgd' o 'adam'.
<li><code>validation_fraction</code>: fracción de datos del conjunto de entrenamiento utilizados como conjunto de validación para el early stopping. Por defecto se utiliza 0.1.
<li><code>n_iter_no_change</code>: número de epochs consecutivas sin mejora que desencadena el early stopping. Por defecto se emplean 10.

 ## **<font color="DeepPink">Problema de clasificación</font>**

<p align="justify">
En este primer ejemplo se muestra cómo, dependiendo de la arquitectura (capas ocultas y tamaño de las mismas), un modelo basado en redes neuronales puede aprender funciones no lineales con gran facilidad. Para evitar problemas de overfitting, es importante identificar la combinación de hiperparámetros que consigue un equilibrio adecuado en el aprendizaje. Se trata de un ejemplo muy sencillo, cuyo objetivo es que se familiarice con la flexibilidad que tienen este tipo de modelos y en cómo realizar una optimización de hiperparámetros mediante grid search y validación cruzada.


```python
# Para análisis y preprocesamiento de datos
import numpy as np
import pandas as pd

# Gráficos
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Configuración warnings
import warnings
warnings.filterwarnings('ignore')

# Preprocesamiento y Modelado
from sklearn.datasets import make_blobs
from sklearn.neural_network import MLPClassifier
from sklearn.neural_network import MLPRegressor
from sklearn.model_selection import RandomizedSearchCV
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import cross_validate
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.compose import make_column_selector as selector
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
import multiprocessing
```

<p align="justify">
Se simulan observaciones en dos dimensiones, pertenecientes a tres grupos, cuya separación no es perfecta.

```python
# Datos simulados con make_blobs
X, y = make_blobs(
        n_samples    = 500,
        n_features   = 2,
        centers      = 3,
        cluster_std  = 1.2,
        shuffle      = True,
        random_state = 0
       )
```

```python
# Gráfico con Plotly
fig = px.scatter(x=X[:,0],
                 y=X[:,1],
                 color=y,
                 template='gridon')
fig.update_layout(title='Datos simulados')
fig.update_coloraxes(showscale=False)
fig.show()
```

Se procede a crear 4 modelos en orden creciente de complejidad (número de neuronas y capas), para comprobar cómo la arquitectura de la red afecta a su capacidad de aprendizaje.

```python
# Modelos: arquitectura de la red

modelo_1 = MLPClassifier(
                hidden_layer_sizes=(5),
                learning_rate_init=0.01,
                solver = 'lbfgs',
                max_iter = 1000,
                random_state = 123
            )

modelo_2 = MLPClassifier(
                hidden_layer_sizes=(10),
                learning_rate_init=0.01,
                solver = 'lbfgs',
                max_iter = 1000,
                random_state = 123
            )

modelo_3 = MLPClassifier(
                hidden_layer_sizes=(20, 20),
                learning_rate_init=0.01,
                solver = 'lbfgs',
                max_iter = 5000,
                random_state = 123
            )

modelo_4 = MLPClassifier(
                hidden_layer_sizes=(50, 50, 50),
                learning_rate_init=0.01,
                solver = 'lbfgs',
                max_iter = 5000,
                random_state = 123
            )
```

<p align="justify">
Se utiliza la función <code>cross_validate()</code> para evaluar cada modelo creado en un conjunto de datos mediante la validación cruzada. El parámetro <code>return_train_score=True</code> indica que también se devolverán las puntuaciones de entrenamiento, esto para evaluar el sobreajuste de los modelos.

```python
cv_results_1 = cross_validate(modelo_1, X, y, return_train_score=True, cv=5)
cv_results_2 = cross_validate(modelo_2, X, y, return_train_score=True, cv=5)
cv_results_3 = cross_validate(modelo_3, X, y, return_train_score=True, cv=5)
cv_results_4 = cross_validate(modelo_4, X, y, return_train_score=True, cv=5)
```

```python
# Crear los datos para la tabla
modelos = ['Modelo 1', 'Modelo 2', 'Modelo 3', 'Modelo 4']
train_scores = [
    cv_results_1['train_score'].mean().round(2),
    cv_results_2['train_score'].mean().round(2),
    cv_results_3['train_score'].mean().round(2),
    cv_results_4['train_score'].mean().round(2)
]
test_scores = [
    cv_results_1['test_score'].mean().round(2),
    cv_results_2['test_score'].mean().round(2),
    cv_results_3['test_score'].mean().round(2),
    cv_results_4['test_score'].mean().round(2)
]

# Crear la figura de la tabla
fig = go.Figure(data=[go.Table(
    header=dict(values=['<b>Modelo</b>', '<b>Train Score</b>', '<b>Test Score</b>'],
                fill_color='deepPink',
                font=dict(color='white', size=14),
                align='center'),
    cells=dict(values=[modelos, train_scores, test_scores],
               fill_color='lavender',
               align='center'))
])

# Actualizar el layout
fig.update_layout(title_text="Resultados de los Modelos")

# Mostrar la figura
fig.show()
```

<p align="justify">
Dado el elevado número de hiperparámetros que tiene los modelos de redes neuronales, la combinación de posibles configuraciones es muy elevada. Sin embargo, se seleccionan aquellos hiperparámetros más influyentes.

```python
# Espacio de búsqueda de cada hiperparámetro
param_distributions = {
    'hidden_layer_sizes': [(10), (10, 10), (20, 20), (50, 50, 50)],
    'alpha': np.logspace(-4, 4, 9),
    'learning_rate_init': [0.001, 0.01, 0.1],
}
```

<p align="justify">
No se realizará la búsqueda de hiperparámetros por grid search cartesiano (todas las combinaciones). En su lugar, se empleará "random grid search", que hace una búsqueda de combinaciones aleatorias.

```python
# Búsqueda por validación cruzada
grid = RandomizedSearchCV(
        estimator  = MLPClassifier(solver = 'lbfgs', max_iter= 2000),
        param_distributions = param_distributions,
        n_iter     = 30, # Número máximo de combinaciones probadas
        scoring    = 'accuracy',
        n_jobs     = multiprocessing.cpu_count() - 1,
        cv         = 3,
        verbose    = 0,
        random_state = 123,
       )
grid.fit(X, y)
```

<p align="justify">
El siguiente código convierte el diccionario de resultados de la validación cruzada de una búsqueda de hiperparámetros (<code>grid.cv_results_</code>) en un DataFrame de pandas y lo ordena según la columna 'rank_test_score', que contiene los rangos de las puntuaciones de las pruebas, facilitando la identificación de los mejores hiperparámetros evaluados.

```python
cv_results = pd.DataFrame(grid.cv_results_).sort_values('rank_test_score')
cv_results.drop(columns=['mean_fit_time',	'std_fit_time',	'mean_score_time',	'std_score_time', 'params', 'split0_test_score', 'split1_test_score', 'split2_test_score', 'std_test_score'], inplace=True)
cv_results.head(10)
```

La combinación de hiperparámetros que mejores resultados consigue es:

```python
grid.best_params_
```

Una vez entrenado y optimizado el modelo de redes neuronales, se realizan las predicción!

```python
grid.predict(X)
```

 ## **<font color="DeepPink">Problema de regresión</font>**

```python
url = ("https://raw.githubusercontent.com/JoaquinAmatRodrigo/Estadistica-machine-learning-python/master/data/SaratogaHouses.csv")
df = pd.read_csv(url)
df.head()
```

<p align="justify">
El set de datos SaratogaHouses del paquete mosaicData de R contiene información sobre el precio de 1728 viviendas situadas en Saratoga County, New York, USA en el año 2006. Además del precio, incluye 15 variables adicionales:
<ul p align="justify">
<li><b>price</b>: precio de la vivienda.
<li><b>lotSize</b>: metros cuadrados de la vivienda.
<li><b>age</b>: antigüedad de la vivienda.
<li><b>landValue</b>: valor del terreno.
<li><b>livingArea</b>: metros cuadrados habitables.
<li><b>pctCollege</b>: porcentaje del vecindario con título universitario.
<li><b>bedrooms</b>: número de dormitorios.
<li><b>firplaces</b>: número de chimeneas.
<li><b>bathrooms</b>: número de cuartos de baño (el valor 0.5 hace referencia a cuartos de baño sin ducha).
<li><b>rooms</b>: número de habitaciones.
<li><b>heating</b>: tipo de calefacción.
<li><b>fuel</b>: tipo de alimentación de la calefacción (gas, electricidad o diesel).
<li><b>sewer</b>: tipo de desagüe.
<li><b>waterfront</b>: si la vivienda tiene vistas al lago.
<li><b>newConstruction</b>: si la vivienda es de nueva construcción.
<li><b>centralAir</b>: si la vivienda tiene aire acondicionado.
</ul>

El objetivo es obtener un modelo capaz de predecir el precio del alquiler.

```python
# Se renombran las columnas para que sean más descriptivas
df.columns = ["precio", "metros_totales", "antiguedad", "precio_terreno",
              "metros_habitables", "universitarios", "dormitorios",
              "chimenea", "banos", "habitaciones", "calefaccion",
              "consumo_calefacion", "desague", "vistas_lago",
              "nueva_construccion", "aire_acondicionado"]
df.head()
```

<p align="justify">
A continuación, se realiza un análisis exploratorio de los datos. Antes de entrenar un modelo predictivo, o incluso antes de realizar cualquier cálculo con un nuevo conjunto de datos, es muy importante realizar una exploración descriptiva de los mismos. Este proceso permite entender mejor qué información contiene cada variable, así como detectar posibles errores.

```python
df.info()
```

```python
px.histogram(df,
             x='precio',
             template='gridon',
             title='Distribución Precio',
             histnorm='density')
```

<p align="justify">
Los modelos de redes neuronales son de tipo no paramétrico, no asumen ningún tipo de distribución de la variable respuesta, por lo tanto, no es necesario que esta siga ninguna distribución concreta. Aun así, siempre es recomendable hacer un estudio mínimo, ya que, a fin de cuentas, es lo que interesa predecir. En este caso, la variable precio tiene una distribución asimétrica con una cola positiva debido a que, unas pocas viviendas, tienen un precio muy superior a la media.

```python
df.describe().T
```

```python
df.describe(include='object').T
```

<p align="justify">
Los modelos de redes neuronales requieren como mínimo de dos tipos de preprocesado: codificación (one hot ecoding) de las variables categóricas y estandarización de las variables numéricas.

```python
y = df.precio
X = df.drop(columns='precio')
```

<p align="justify">
La función de scikit-learn <code>make_column_selector</code> permite seleccionar columnas según el tipo de datos que posea en forma automatizada, sin tener que hacerlo manual.
<br><br>
Se crea el selector pasando el tipo de datos, luego se aplica al conjunto de datos. El resultado será una lista de nombres de las columnas que tienen el tipo de datos indicado.

```python
numerical_columns_selector = selector(dtype_exclude=object)
categorical_columns_selector = selector(dtype_include=object)
```

```python
numerical_columns = numerical_columns_selector(X)
categorical_columns = categorical_columns_selector(X)
```

```python
numerical_columns
```

```python
categorical_columns
```

<p align="justify">
<code>ColumnTransformer()</code> crea un objeto ColumnTransformer que aplica una codificación one-hot a las columnas categóricas y una estandarización a las columnas numéricas del DataFrame, permitiendo preprocesar adecuadamente diferentes tipos de datos antes de entrenar un modelo.

```python
preprocessor = ColumnTransformer([
    ('one-hot-encoder', OneHotEncoder(handle_unknown='ignore'), categorical_columns),
    ('standard-scaler', StandardScaler(), numerical_columns)])
```

```python
preprocessor
```

<p align="justify">
Se crea un pipeline que encadena <code>preprocesador</code> y un modelo de regresión. El modelo de regresión es un <code>MLPRegressor</code> (Multi-Layer Perceptron Regressor).

```python
model_pipeline = make_pipeline(preprocessor, MLPRegressor(solver = 'lbfgs', max_iter= 1000, random_state = 123))
```

<p align="justify">
Se realiza la validación cruzada del modelo utilizando 5 pliegues (folds) y evaluando el rendimiento con la métrica de error absoluto medio negativo (neg_mean_absolute_error).

```python
cv_results = cross_validate(model_pipeline, X, y, cv=5, n_jobs = 2, scoring='neg_mean_absolute_error')
```

```python
-(cv_results['test_score'].mean().round(2))
```

```python
# Espacio de búsqueda de cada hiperparámetro
param_grid = {
    'mlpregressor__hidden_layer_sizes': [(10), (20), (10, 10)],
    'mlpregressor__alpha': np.logspace(-4, 4, 9),
    'mlpregressor__learning_rate_init': [0.001, 0.01]}
```

<p align="justify">
<code>GridSearchCV</code> realiza una búsqueda exhaustiva de hiperparámetros para el pipeline, utilizando el conjunto de parámetros <code>param_grid</code>. La búsqueda se realiza utilizando todos los núcleos de CPU disponibles menos uno (<code>n_jobs=multiprocessing.cpu_count() - 1</code>), con una validación cruzada de 3 pliegues, y se evalúan las combinaciones de parámetros usando la métrica de error absoluto medio negativo.
<br><br>
Finalmente, <code>model_grid_search.fit(X, y)</code> ajusta este objeto a los datos X e y, encontrando las mejores combinaciones de parámetros y entrenando el modelo con ellas.

```python
# Búsqueda por validación cruzada
grid = GridSearchCV(estimator       = model_pipeline,
                    param_grid      = param_grid,
                    n_jobs          = multiprocessing.cpu_count() - 1,
                    cv              = 3,
                    scoring         = 'neg_mean_absolute_error')
grid.fit(X, y)
```

```python
cv_results = pd.DataFrame(grid.cv_results_).sort_values('rank_test_score')
cv_results.drop(columns=['mean_fit_time',	'std_fit_time',	'mean_score_time',	'std_score_time', 'params', 'split0_test_score', 'split1_test_score', 'split2_test_score', 'std_test_score'], inplace=True)
cv_results.head(10)
```

```python
grid.best_params_
```

```python
-grid.best_score_.round(1)
```

```python
grid.predict(X)
```
