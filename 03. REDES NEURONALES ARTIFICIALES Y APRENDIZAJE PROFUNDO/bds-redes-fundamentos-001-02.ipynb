## **<font color="DeepPink">Proceso de entrenamiento de una Red Neuronal Artificial</font>**

<ol p align="justify">
<li><b>Inicialización</b>: los pesos y sesgos de las neuronas se inicializan aleatoriamente.
<br><br>
<li><b>Propagación hacia Adelante (Forward Propagation)</b>: los datos de entrada se pasan a través de la red capa por capa. Cada neurona en una capa procesa la entrada, aplican los pesos y sesgos, para luego pasar por una función de activación y producir una salida. Este proceso continúa hasta que la capa de salida (<i>output layer</i>) produce la predicción de la red.
<br><br>
<li><b>Cálculo de Pérdida (Loss Calculation)</b>: la salida de la red se compara con los valores objetivo reales (etiqueta) usando una <b>función de pérdida</b> (como el error cuadrático medio para tareas de regresión o la pérdida de <a href = "https://en.wikipedia.org/wiki/Cross-entropy">entropía cruzada</a> para tareas de clasificación). La función de pérdida calcula la diferencia (error) entre los valores predichos y los valores reales.
<br><br>
<li><b>Propagación hacia Atrás (Backpropagation)</b>: el error se propaga de vuelta a través de la red desde la capa de salida hasta la capa de entrada. Se calculan los gradientes de la función de pérdida con respecto a los pesos y sesgos usando la regla de la cadena del cálculo. Estos gradientes indican cuánto cambiaría la pérdida respecto a pequeños cambios en los pesos y sesgos.
<br><br>
<li><b>Actualización de pesos y sesgos</b>: los pesos y sesgos de la red se actualizan en la dirección que reduce la pérdida, típicamente usando un algoritmo de optimización como el descenso de gradiente. La regla de actualización de pesos generalmente es:
$$
w_{ \text{nuevo}} = w_{ \text{viejo}} - \eta \frac{\partial J}{\partial w}
$$


donde $\eta$ es la tasa_de_aprendizaje (<code>learning_rate</code>), este es un hiperparámetro que controla el tamaño de la actualización. Y por ejemplo, como se observa en el siguiente gráfico, $\frac{\partial J}{\partial w}$ sería el gradiente de la función de pérdida con respecto al peso ($w$).

<li><b>Iteración</b>: los pasos 2 a 5 se repiten durante muchas iteraciones (epochs) hasta que el rendimiento de la red en los datos de entrenamiento sea satisfactorio o la pérdida converja a un valor mínimo.


## **<font color="DeepPink">Fundamento matemático de Backpropagation</font>**

<ol p align="justify">
Backpropagation se basa en la regla de la cadena del cálculo para calcular el gradiente de la función de pérdida respecto a cada peso y sesgo. Para una red simple, esto implica:
<ul p align="justify">
<li>Calcular la derivada parcial de la pérdida respecto a la salida de cada neurona.
<li>Calcular la derivada parcial de la activación (salida) de cada neurona respecto a su suma ponderada.
<li>Calcular la derivada parcial de la suma ponderada de la neurona respecto a sus pesos y sesgos.
<li>Usar estas derivadas parciales para aplicar la regla de la cadena y computar el gradiente de la pérdida respecto a cada peso y sesgo.
</ul>
<p align="justify">
Si necesita repasar conceptos de derivada parcial le recomedamos la siguiente <a href="https://www.khanacademy.org/math/multivariable-calculus/multivariable-derivatives/partial-derivative-and-gradient-articles/a/introduction-to-partial-derivatives">página</a>



Consideremos una red neuronal muy simple:

Calculamos el error:
$$E = \frac{1}{2} (T - a_2)^2$$
y actualizamos $w_2, b_2, w_1$ y $b_1$





Generalizando el error:
$$E = \frac{1}{2m} \sum_{i=1}^{m} (T_i - a_{2,i})^2$$

Actualizamos $w_2$:
$$
w_{2, \text{nuevo}} = w_{2, \text{viejo}} - \eta \frac{\partial E}{\partial w_2}
$$
$E = \frac{1}{2} (T - a_2)^2 \rightarrow \frac{\partial E}{\partial a_2}$

$a_2 = f(z_2) = \frac{1}{1 + e^{-z_2}} \rightarrow \frac{\partial a_2}{\partial z_2}$

$z_2 = a_1w_2+b_2 \rightarrow \frac{\partial z_2}{\partial w_2}$

$\frac{\partial E}{\partial w_2}=\frac{\partial E}{\partial a_2}\frac{\partial a_2}{\partial z_2}\frac{\partial z_2}{\partial w_2}$

$\frac{\partial E}{\partial w_2}=(-(T-a_2))(a_2(1-a_2))(a_1)$

$$
w_{2, \text{nuevo}} = w_{2, \text{viejo}} - \eta (-(T-a_2))(a_2(1-a_2))(a_1)
$$

Actualizamos $b_2$:


$\frac{\partial E}{\partial b_2}=\frac{\partial E}{\partial a_2}\frac{\partial a_2}{\partial z_2}\frac{\partial z_2}{\partial b_2}$

$\frac{\partial E}{\partial b_2}=(-(T-a_2))(a_2(1-a_2))1$

$$
b_{2, \text{nuevo}} = b_{2, \text{viejo}} - \eta (-(T-a_2))(a_2(1-a_2))1
$$

Actualizamos $w_1$:
$$
w_{1, \text{nuevo}} = w_{1, \text{viejo}} - \eta \frac{\partial E}{\partial w_1}
$$
$E = \frac{1}{2} (T - a_2)^2 \rightarrow \frac{\partial E}{\partial a_2}$

$a_1 = f(z_1) = \frac{1}{1 + e^{-z_1}} \rightarrow \frac{\partial a_1}{\partial z_1}$

$a_2 = f(z_2) = \frac{1}{1 + e^{-z_2}} \rightarrow \frac{\partial a_2}{\partial z_2}$

$z_1 = x_1w_1+b_1 \rightarrow \frac{\partial z_1}{\partial w_1}$

$z_2 = a_1w_2+b_2 \rightarrow \frac{\partial z_2}{\partial a_1}$

$\frac{\partial E}{\partial w_1}=\frac{\partial E}{\partial a_2}\frac{\partial a_2}{\partial z_2}\frac{\partial z_2}{\partial a_1}\frac{\partial a_1}{\partial z_1}\frac{\partial z_1}{\partial w_1}$

$\frac{\partial E}{\partial w_1}=(-(T-a_2))(a_2(1-a_2))(w_2)a_1(1-a_1)x_1$

$$
w_{1, \text{nuevo}} = w_{1, \text{viejo}} - \eta (-(T-a_2))(a_2(1-a_2))(w_2)a_1(1-a_1)x_1
$$

Actualizamos $b_1$:
$$
b_{1, \text{nuevo}} = b_{1, \text{viejo}} - \eta \frac{\partial E}{\partial b_1}
$$


$\frac{\partial E}{\partial b_1}=\frac{\partial E}{\partial a_2}\frac{\partial a_2}{\partial z_2}\frac{\partial z_2}{\partial a_1}\frac{\partial a_1}{\partial z_1}\frac{\partial z_1}{\partial b_1}$

$\frac{\partial E}{\partial b_1}=(-(T-a_2))(a_2(1-a_2))(w_2)a_1(1-a_1)1$

$$
b_{1, \text{nuevo}} = b_{1, \text{viejo}} - \eta (-(T-a_2))(a_2(1-a_2))(w_2)a_1(1-a_1)1
$$

## **<font color="DeepPink">Creación de una red neuronal con Backward Propagation (Backpropagation)</font>**

<p align="justify">
A continuación se desarrollará un ejemplo extraido del artículo <a href="https://machinelearningmastery.com/implement-backpropagation-algorithm-scratch-python/">How to Code a Neural Network with Backpropagation In Python (from scratch)</a> de Jason Brownlee.

````python
import numpy as np
from random import seed
from random import random
#from math import exp
````

### **<font color="DeepPink">1. Inicialización de la red</font>**

<p align="justify">
Empecemos por algo sencillo, la creación de una nueva red con una capa oculta.
<br><br>
La función llamada <code>inicialize_network</code> crea una nueva red neuronal lista para entrenar. Acepta tres parámetros, el número de entradas (<code>n_inputs</code>), el número de neuronas a tener en la capa oculta (<code>n_hidden</code>) y el número de salidas (<code>n_outputs</code>).
<br><br>
Para la <b>capa oculta</b> (<code>hidden_layer</code>) se crean <code>n_hidden</code> (n neuronas ocultas) y cada neurona en la capa oculta tiene <code>n_inputs + 1</code> pesos, uno para cada columna de entrada en un conjunto de datos y uno adicional para el sesgo.
<br><br>
La <b>capa de salida</b> (<code>output_layer</code>) que se conecta a la capa oculta tiene <code>n_outputs</code> (n neuronas de salida), cada una con <code>n_hidden + 1</code> pesos.

````python
def initialize_network(n_inputs, n_hidden, n_outputs):
    network = list()
    hidden_layer = [{'weights':[random() for i in range(n_inputs + 1)]} for i in range(n_hidden)]
    network.append(hidden_layer)
    output_layer = [{'weights':[random() for i in range(n_hidden + 1)]} for i in range(n_outputs)]
    network.append(output_layer)
    return network
````

<p align="justify">
Se crea una red neuronal con 1 capa oculta y 1 neurona en dicha capa.
La capa oculta tiene 2 pesos (weights) y el sesgo (bias), mientras que la capa de salida tiene 2 neuronas, cada una con 1 peso más el sesgo.


````python
seed(42)
network = initialize_network(2, 1, 2)
for layer in network:
    print(layer)
````

### **<font color="DeepPink">2. Forward Propagation</font>**

A continuación se computa la salida mediante **Forward Propagation**.

Primero realizamos la suma ponderada de las entradas.

$$z=w_1x_1+w_2x_2+...+w_nx_n+b$$

$$z= \sum^n_{i=1} x_i w_i + b$$

A continuación se realiza una implementación de esta fórmula en una función denominada `compute_weighted_sum()`. La función supone que el sesgo es el último peso en la lista de pesos.

````python
# Se calcula la suma ponderada a partir de los pesos y las entradas
def compute_weighted_sum(weights, inputs):
    weighted_sum = weights[-1]
    for i in range(len(weights)-1):
        weighted_sum += weights[i] * inputs[i]
    return weighted_sum
````

<p align="justify">
Una vez que realizamos la suma ponderada, es necesario realizar la activación para ver cuál es realmente la salida de la neurona.
<br><br>
Las <b>funciones de activación</b> controlan en gran medida que información se propaga desde una capa a la siguiente (<b>forward propagation</b>). Estas funciones convierten el valor de la combinación de los entradas, pesos y sesgo, en un nuevo valor. Es gracias combinar funciones de activación no lineales con múltiples capas, que los modelos de redes son capaces de aprender relaciones no lineales.
<br><br>
Existen diferentes <b>funciones de activación</b>. La más impleada es la función sigmoide (<b>Sigmoid</b>), sin embargo existen otras como Tangente hiperbólica (<b>Tanh</b>) o Rectified linear unit (<b>ReLU</b>) para activar salidas.


**Función ReLU**
$$a = f(z) = max(z,0)$$
<p align="justify">
La función de activación ReLU aplica una transformación no lineal muy simple, activa la neurona solo si el valor de entrada está por encima de cero. Mientras el valor de entrada está por debajo de cero, el valor de salida es cero, pero cuando es superior de cero, el valor de salida aumenta de forma lineal con el de entrada.

**Función Sigmoid**
$$
a = f(z) = \frac{1}{1 + e^{-z}}
$$
<p align="justify">
La función de activación Sigmoid tiene forma de S, también se llama función logística. Puede tomar cualquier valor de entrada y producir un número entre 0 y 1 en una curva S. También es una función cuya derivada se puede calcular fácilmente, lo cual será necesario más adelante al propagar el error hacia atrás (Backpropagation).

**Función Tanh**
$$a = f(z) = \frac{1 - e^{-2z}}{1 + e^{-2z}}$$
<p align="justify">
La función de activación Tanh, se comporta de forma similar a la función Sigmoid, pero su salida está acotada en el rango (-1, 1).

````python
# Se utiliza la función sigmoide como función de activación
def activation(weighted_sum):
    return 1.0 / (1.0 + np.exp(-weighted_sum))
````

<p align="justify">
Se trabaja en cada capa de la red calculando las salidas para cada neurona. Todas las salidas de una capa se convierten en entradas para las neuronas de la siguiente capa.
<br><br>
A continuación, se muestra una función llamada <code>forward_propagate()</code> que implementa la propagación hacia adelante de una fila de datos del conjunto de datos con la red neuronal.
<br><br>
Se puede ver que el valor de salida de una neurona se almacena en la neurona con el nombre <code>output</code>. También se puede observar que se recopilan las salidas de una capa en una matriz denominada <code>new_inputs</code>, la cual se convierte en las entradas de la matriz y se utiliza como entradas para la siguiente capa.
<br><br>
Esta función devuelve las salidas de la última capa, también llamada capa de salida. A continuación, se muestra el resultado de la propagación hacia adelante de la red neuronal con una neurona oculta que espera 2 valores de entrada y una capa de salida con dos neuronas.

````python
def forward_propagate(network, row):
    inputs = row
    for layer in network:
        new_inputs = []
        for neuron in layer:
            weighted_sum = compute_weighted_sum(neuron['weights'], inputs)      # Se calcula la suma ponderada y se almacena en una variable
            neuron['output'] = activation(weighted_sum)                         # Se pasa la suma ponderada calculada a la función de activación
            new_inputs.append(neuron['output'])
        inputs = new_inputs
    return inputs
````

````python
# Se prueba el resultado de una red neuronal ("network")
network = [[{'weights': [0.13436424411240122, 0.8474337369372327, 0.763774618976614]}],
 [{'weights': [0.2550690257394217, 0.49543508709194095]}, {'weights': [0.4494910647887381, 0.651592972722763]}]]
row = [1, 0]                                                                    # Valores iniciales de las entradas
output = forward_propagate(network, row)
print(output)
````

<p align="justify">
Debido a que la capa de salida tiene dos neuronas, se obtiene una lista de dos números (predicciones).
<br><br>
Los valores de salida reales no tienen sentido por ahora, pero a continuación se comenzará a aprender cómo hacer que los pesos en las neuronas sean más útiles.

### **<font color="DeepPink">3. Backpropagation</font>**

<p align="justify">
El algoritmo de backpropagation lleva el nombre de la forma en que se entrenan los pesos.
<br><br>
El error se calcula entre las salidas esperadas (valores reales) y las salidas propagadas hacia adelante desde la red (predicción). Luego, estos errores se propagan hacia atrás a través de la red desde la capa de salida a la capa oculta, asignando "responsabilidades" por el error y actualizando los pesos.
<BR><BR>
Aquí se realizan 2 pasos:

*   **Transferencia Derivada**.
*   **Error de retropropagación**.

<p align="justify">
La función <CODE>transfer_derivative()</CODE> calcula la derivada de la función de activación Sigmoid:
<br><br>
$$f'(z)=f(z).(1-f(z))$$


````python
# La función para la derivada de la función de activación (Sigmoid) será:
def transfer_derivative(output):
    return output * (1.0 - output)
````

Calculamos el error de cada neurona de salida, esto nos dará nuestra señal de error para propagarla hacia atrás a través de la red. El error de una neurona determinada se puede calcular de la siguiente manera:


```
error = (output - expected) * transfer_derivative(output)
```
<p align="justify">
Este cálculo de error se utiliza para las neuronas de la capa de salida. En la capa oculta, las cosas son un poco más complicadas.
<br><br>
La señal de error de una neurona en la capa oculta se calcula como el error ponderado de cada neurona en la capa de salida. Piense en el error que viaja a lo largo de los pesos de la capa de salida hasta las neuronas en la capa oculta.
<br><br>
La señal de error retropropagada se acumula y luego se utiliza para determinar el error de la neurona en la capa oculta, de la siguiente manera:



```
error = (weight_k * error_j) * transfer_derivative(output)
```
<p align="justify">
Donde <code>error_j</code> es la señal de error de la neurona j en la capa de salida, <code>weight_k</code> es el peso que conecta la neurona k a la neurona actual y <code>output</code> es la salida de la neurona actual.
<br><br>
A continuación se muestra una función llamada <code>backward_propagate_error()</code> que implementa este procedimiento:



````python
# Se define la función para el error de retropropagación
def backward_propagate_error(network, expected):
    for i in reversed(range(len(network))):
        layer = network[i]
        errors = list()
        if i != len(network)-1:
            for j in range(len(layer)):
                error = 0.0
                for neuron in network[i + 1]:
                    error += (neuron['weights'][j] * neuron['delta'])
                    errors.append(error)
        else:
            for j in range(len(layer)):
                neuron = layer[j]
                errors.append(neuron['output'] - expected[j])
        for j in range(len(layer)):
            neuron = layer[j]
            neuron['delta'] = errors[j] * transfer_derivative(neuron['output'])
````

<p align="justify">
Se puede observar que la señal de error calculada para cada neurona se almacena con el nombre <code>delta</code>. Las capas de la red se iteran en orden inverso, comenzando en la salida y trabajando hacia atrás. Esto garantiza que las neuronas en la capa de salida tengan valores <code>delta</code> calculados primero, que las neuronas en la capa oculta pueden usar en la iteración posterior. Se eligió el nombre <code>delta</code> para reflejar el cambio que implica el error en la neurona (por ejemplo, el peso delta).
<br><br>
Se puede ver que la señal de error de las neuronas en la capa oculta se acumula a partir de las neuronas en la capa de salida, donde el número de neurona oculta j también es el índice del peso de la neurona en la capa de salida neuron <code>['weights'][j]</code>.
<br><br>
Al ejecutar el ejemplo, se imprime la red una vez finalizada la retropropagación del error. Puede ver que los valores de error se calculan y almacenan en las neuronas de la capa de salida y la capa oculta.

````python
# Se prueba el resultado
network = [[{'output': 0.7105668883115941, 'weights': [0.13436424411240122, 0.8474337369372327, 0.763774618976614]}],
[{'output': 0.6213859615555266, 'weights': [0.2550690257394217, 0.49543508709194095]}, {'output': 0.6573693455986976, 'weights': [0.4494910647887381, 0.651592972722763]}]]
expected = [0, 1]
backward_propagate_error(network, expected)
for layer in network:
    print(layer)
````

### **<font color="DeepPink">4. Entrenamiento de la red</font>**

<p align="justify">
La red se entrena utilizando el descenso del gradiente. Esto implica múltiples iteraciones de exposición de un conjunto de datos de entrenamiento a la red y, para cada fila de datos, propagar hacia adelante las entradas, propagar hacia atrás el error y actualizar los pesos de la red.
<br><br>
Aquí realizamos 2 pasos:

*   **Actualizar pesos**.
*   **Entrenar la red**.

<p align="justify">
Una vez que se calculan los errores para cada neurona de la red a través del método de retropropagación mencionado anteriormente, se pueden utilizar para actualizar los pesos.
<br><br>
Los pesos de la red se actualizan de la siguiente manera:

```
weight = weight - learning_rate * error * input
```
<p align="justify">
Donde <code>weight</code> es un peso dado, <code>learning_rate</code> es un parámetro que debe especificar, <code>error</code> es el error calculado por el procedimiento de retropropagación para la neurona e <code>input</code> es el valor de entrada que causó el error.
<br><br>
Se puede utilizar el mismo procedimiento para actualizar el peso del sesgo, excepto que no hay ningún término de entrada o la entrada es el valor fijo de 1.
<br><br>
La tasa de aprendizaje (<code>learning_rate</code>) controla cuánto se debe cambiar el peso para corregir el error. Por ejemplo, un valor de 0.1 actualizará el peso en un 10 % de la cantidad en la que posiblemente podría actualizarse. Se prefieren tasas de aprendizaje pequeñas que provoquen un aprendizaje más lento a lo largo de una gran cantidad de iteraciones de entrenamiento. Esto aumenta la probabilidad de que la red encuentre un buen conjunto de pesos en todas las capas en lugar del conjunto de pesos más rápido que minimiza el error (lo que se denomina convergencia prematura).
<br><br>
A continuación se muestra una función llamada <code>update_weights()</code> que actualiza los pesos de una red dada una fila de entrada de datos, una tasa de aprendizaje y supone que ya se ha realizado una propagación hacia adelante y hacia atrás.


````python
# Hacemos el update de los weights
def update_weights(network, row, l_rate):
    for i in range(len(network)):
        inputs = row[:-1]
        if i != 0:
            inputs = [neuron['output'] for neuron in network[i - 1]]
        for neuron in network[i]:
            for j in range(len(inputs)):
                neuron['weights'][j] -= l_rate * neuron['delta'] * inputs[j]
            neuron['weights'][-1] -= l_rate * neuron['delta']
````

<p align="justify">
Ahora que sabemos cómo actualizar los pesos de la red, veamos cómo podemos hacerlo iterativamente.
<br><br>
Esto implica primero realizar un bucle durante un número fijo de epochs (iteraciones) y dentro de cada epoch actualizar la red para cada fila del conjunto de datos de entrenamiento.
<br><br>
A continuación se muestra una función que implementa el entrenamiento (<code>train_network()</code>) de una red neuronal ya inicializada (<code>network</code>) con un conjunto de datos de entrenamiento determinado (<code>train</code>), una tasa de aprendizaje (<code>l_rate</code>), un número fijo de epochs (<code>n_epochs</code>) y un número esperado de valores de salida (<code>n_outputs</code>).


````python
# Finalmente, se entrena el modelo para un número dado de epochs
def train_network(network, train, l_rate, n_epoch, n_outputs):
    for epoch in range(n_epoch):
        sum_error = 0
        for row in train:
            outputs = forward_propagate(network, row)
            expected = [0 for i in range(n_outputs)]
            expected[row[-1]] = 1
            sum_error += sum([(expected[i]-outputs[i])**2 for i in range(len(expected))])
            backward_propagate_error(network, expected)
            update_weights(network, row, l_rate)
        print('>epoch=%d, lrate=%.3f, error=%.3f' % (epoch, l_rate, sum_error))
````

<p align="justify">
A continuación se muestra un pequeño conjunto de datos artificiales que podemos usar para probar el entrenamiento de la red neuronal.
<br><br>
La red se entrenará durante 20 epochs con una tasa de aprendizaje de 0.5, que es alta porque se está entrenando para muy pocas iteraciones.

````python
# Testeamos la red construida de esta forma para un ejemplo particular.
seed(1)
dataset = [[2.7810836,2.550537003,0],
 [1.465489372,2.362125076,0],
 [3.396561688,4.400293529,0],
 [1.38807019,1.850220317,0],
 [3.06407232,3.005305973,0],
 [7.627531214,2.759262235,1],
 [5.332441248,2.088626775,1],
 [6.922596716,1.77106367,1],
 [8.675418651,-0.242068655,1],
 [7.673756466,3.508563011,1]]
n_inputs = len(dataset[0]) - 1
n_outputs = len(set([row[-1] for row in dataset]))
network = initialize_network(n_inputs, 2, n_outputs)
train_network(network, dataset, 0.5, 20, n_outputs)
for layer in network:
    print(layer)
````

<p align="justify">
Al ejecutar el ejemplo primero se imprime el error de suma al cuadrado de cada epoch de entrenamiento. Se aprecia una tendencia de disminución de este error con cada epoch.
<br><br>
Una vez entrenada, la red se imprime y muestra los pesos aprendidos. También se encuentran en la red valores de salida y delta que se pueden ignorar.
<br><br>
Una vez entrenada una red, podemos usarla para hacer predicciones!

## **<font color="DeepPink">Resumen</font>**



1.   Se inicializan los pesos y bias.
2.   Proceso iterativo de los siguientes pasos:
    1. Calcular la salida de la red usando Forward Propagation.
    2. Calcular el error entre el valor real y la salida estimada (predicción).
    3. Actualizar los pesos y bias mediante Backpropagation.
    4. Repetir los 3 pasos anteriores hasta que el número de iteraciones o epochs se alcance o hasta que el error esté por debajo de cierto umbral.
