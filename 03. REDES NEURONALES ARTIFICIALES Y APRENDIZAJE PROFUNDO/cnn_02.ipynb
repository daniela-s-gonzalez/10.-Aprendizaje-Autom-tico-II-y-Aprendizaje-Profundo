```python
fmnist = tf.keras.datasets.fashion_mnist
(training_images, training_labels), (test_images, test_labels) = fmnist.load_data()
```

```python
training_images = training_images / 255.0
test_images = test_images / 255.0
```

```python
# Definimos los nombres de las clases de Fashion-MNIST
class_names = ["T-shirt/top",   # 0
               "Trouser",       # 1
               "Pullover",      # 2
               "Dress",         # 3
               "Coat",          # 4
               "Sandal",        # 5
               "Shirt",         # 6
               "Sneaker",       # 7
               "Bag",           # 8
               "Ankle boot"]    # 9
```

Elegimos 6000 imágenes para entrenar y 2000 para testear

```python
# Seleccionamos solo una parte del dataset
training_images = training_images[:6000]
training_labels = training_labels[:6000]

test_images = test_images[:2000]
test_labels = test_labels[:2000]
```

 ## **<font color="DeepPink">Definimos la DNN (red neuronal totalmente conectada (fully connected))</font>**


<div style="font-family: Arial, sans-serif; line-height: 1.6; color: #202124; padding: 18px; border-left: 5px solid #1A73E8; background-color: #f8f9fa; border-radius: 6px;">
  <h2 style="color: #1a73e8; margin-top: 0; border-bottom: 1px solid #dadce0; padding-bottom: 8px;">Conceptos Clave del Modelo</h2>
  <ul style="list-style-type: none; padding-left: 0; margin: 15px 0;">
    <li style="margin-bottom: 14px;">
      <strong style="color: #1a73e8; font-size: 1.05em;">Clasificación Supervisada Multiclase</strong>
      <span style="display: block; margin-top: 4px; color: #3c4043;">
        El objetivo principal del modelo es asignar cada imagen a una categoría específica dentro de un conjunto predefinido de opciones mediante datos de entrenamiento etiquetados.
      </span>
      <ul style="margin-top: 6px; padding-left: 20px; color: #3c4043;">
        <li><strong>Multiclase:</strong> Existen más de dos categorías posibles (en Fashion-MNIST son 10 clases distintas: camisetas, pantalones, vestidos, zapatos, etc.). El modelo no realiza una decisión simple de "sí/no" (binaria), sino que evalúa múltiples opciones simultáneamente.</li>
        <li><strong>Supervisado:</strong> El algoritmo aprende a partir de un conjunto de datos donde cada imagen ya tiene asignada su respuesta correcta (etiqueta).</li>
      </ul>
    </li>
    <li style="margin-bottom: 0;">
      <strong style="color: #1a73e8; font-size: 1.05em;">Capa de Entrada (Input Layer)</strong>
      <span style="display: block; margin-top: 4px; color: #3c4043;">
        Se define mediante la propiedad <code style="background-color: #e8f0fe; color: #1a73e8; padding: 2px 6px; border-radius: 4px; font-weight: bold;">shape=(28, 28, 1)</code>, una tupla que especifica las dimensiones exactas de cada muestra de entrada:
      </span>
      <ul style="margin-top: 6px; padding-left: 20px; color: #3c4043;">
        <li><code style="background-color: #e8eaed; padding: 2px 5px; border-radius: 3px;">28</code> → Alto de la imagen (28 píxeles).</li>
        <li><code style="background-color: #e8eaed; padding: 2px 5px; border-radius: 3px;">28</code> → Ancho de la imagen (28 píxeles).</li>
        <li><code style="background-color: #e8eaed; padding: 2px 5px; border-radius: 3px;">1</code> → Número de canales de color (1 canal al tratarse de imágenes en escala de grises).</li>
      </ul>
    </li>
  </ul>
</div>

```python
# Definimos el Modelo
model = tf.keras.models.Sequential([
    tf.keras.Input(shape=(28,28,1)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation=tf.nn.relu),
    tf.keras.layers.Dense(10, activation=tf.nn.softmax)
])
```

```python
# Compilamos el Modelo
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
# Enrenamos el Modelo
history = model.fit(training_images, training_labels,
                    epochs=10,
                    validation_data=(test_images, test_labels))
```

  <p style="margin-top: 10px;">
    La Red Neuronal Densa (<strong>DNN</strong>) logra un rendimiento sólido en el conjunto de datos <strong>Fashion-MNIST</strong>, alcanzando aproximadamente un <strong>~90%</strong> de <em>accuracy</em> en entrenamiento y un <strong>~85%</strong> en validación.
  </p>

    <li style="margin-bottom: 12px;">

      <span style="display: block; margin-top: 3px; color: #3c4043;">
        El mejor balance se alcanza entre la <strong>Época 7 y 8</strong> (<em>Epoch 7–8</em>), donde el modelo logra su máximo nivel de generalización.
      </span>
    </li>
    <li style="margin-bottom: 0;">
      <strong style="color: #1e8e3e;">Detección de Sobreajuste (Overfitting):</strong>

    <ul style="margin-top: 6px; margin-bottom: 0; padding-left: 20px;">
      <li>Implementar un callback como <code style="background-color: #ffffff; padding: 2px 5px; border-radius: 3px; border: 1px solid #ceedd5;">EarlyStopping</code> o limitar directamente el número de épocas a 7-8 para detener el entrenamiento antes del sobreajuste.</li>
      <li>Para superar la barrera del 85% de exactitud, lo ideal es migrar a una <strong>Red Neuronal Convolucional (CNN)</strong>, la cual aprovecha de forma superior las relaciones y estructuras espaciales de las imágenes.</li>
    </ul>
  </div>
</div>

```python
# Graficamos la evolución de la accuracy y la loss
plt.figure(figsize=(12,5))

# Accuracy
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Entrenamiento')
plt.plot(history.history['val_accuracy'], label='Validación')
plt.title('Evolución de la Accuracy')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

# Loss
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Entrenamiento')
plt.plot(history.history['val_loss'], label='Validación')
plt.title('Evolución de la Loss')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.show()
```

 ## **<font color="DeepPink">Pasando de una DNN a una CNN</font>**


<div style="font-family: Arial, sans-serif; line-height: 1.6; color: #202124; padding: 18px; border-left: 5px solid #1A73E8; background-color: #f8f9fa; border-radius: 6px;">
  <h2 style="color: #1a73e8; margin-top: 0; border-bottom: 1px solid #dadce0; padding-bottom: 8px;">Modelo Mejorado: Red Neuronal Convolucional (CNN)</h2>
  <h3 style="color: #1a73e8; margin-top: 16px; border-bottom: 1px solid #dadce0; padding-bottom: 4px;">Componentes de la Arquitectura</h3>
  <ul style="list-style-type: none; padding-left: 0; margin: 12px 0;">
    <li style="margin-bottom: 12px;">
      <strong style="color: #1a73e8;">Convoluciones (<code>Conv2D</code>):</strong>
      <span style="display: block; margin-top: 3px; color: #3c4043;">
        Utiliza filtros de <code style="background-color: #e8eaed; padding: 2px 5px; border-radius: 3px;">3 × 3</code> que recorren la imagen para detectar patrones locales (bordes, curvas y texturas). Al aplicar 64 filtros, el modelo genera 64 mapas de características distintos.
      </span>
    </li>
    <li style="margin-bottom: 12px;">
      <strong style="color: #1a73e8;">Capa de Reducción (<code>MaxPooling2D</code>):</strong>
      <span style="display: block; margin-top: 3px; color: #3c4043;">
        Reduce la dimensión espacial de los mapas de características (por ejemplo, de <code style="background-color: #e8eaed; padding: 2px 5px; border-radius: 3px;">26 × 26</code> a <code style="background-color: #e8eaed; padding: 2px 5px; border-radius: 3px;">13 × 13</code>). Conserva los rasgos más relevantes, descarta el ruido, mejora la generalización y disminuye el costo computacional.
      </span>
    </li>
    <li style="margin-bottom: 0;">
      <strong style="color: #1a73e8;">Clasificación Final (Capas Densas):</strong>
      <span style="display: block; margin-top: 3px; color: #3c4043;">
        Tras extraer las características espaciales, la información se aplana (<code>Flatten</code>) y se conecta a capas densas. La capa final utiliza la función de activación <code style="background-color: #e8eaed; padding: 2px 5px; border-radius: 3px;">softmax</code> para devolver la distribución de probabilidad entre las 10 clases.
      </span>
    </li>
  </ul>
  <h3 style="color: #1a73e8; margin-top: 20px; border-bottom: 1px solid #dadce0; padding-bottom: 4px;">Ventajas Clave de la CNN frente a la DNN</h3>
  <ul style="padding-left: 20px; color: #3c4043; margin-top: 10px; margin-bottom: 15px;">
    <li><strong>Captura de información espacial:</strong> Reconoce patrones locales en 2D que una DNN tradicional pierde al aplanar la imagen al inicio.</li>
    <li><strong>Mayor capacidad de generalización:</strong> La reducción mediante pooling disminuye el riesgo de sobreajuste.</li>
    <li><strong>Mejor rendimiento:</strong> Alcanza una precisión significativamente superior en conjuntos de datos de visión por computadora como Fashion-MNIST.</li>
    <li><strong>Escalabilidad:</strong> La arquitectura permite extenderse fácilmente hacia imágenes de mayor tamaño o complejidad.</li>
  </ul>
  <div style="margin-top: 15px; padding: 12px 15px; background-color: #e8f0fe; border-radius: 5px; font-size: 0.95em; color: #174ea6; border: 1px solid #d2e3fc;">
    💡 <strong>En resumen:</strong> La mejora fundamental consiste en evolucionar de un modelo que procesa <em>"píxeles aislados"</em> (DNN) a uno que aprende <em>"patrones visuales jerárquicos"</em> (CNN), optimizando sustancialmente la capacidad de clasificación del modelo.
  </div>
</div>

```python
# Definimos el Modelo
model = tf.keras.models.Sequential([

    # Agregamos convolutions y max pooling
    tf.keras.Input(shape=(28,28,1)),
    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2, 2),
    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2,2),

    # Agrgamos las mismas capas que antes
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])
```

```python
#Imprimimos el Resumen del Modelo
model.summary()
```

```python
# Usamos la misma configuración
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
# Entrenamos el Modelo
history = model.fit(training_images, training_labels,
                    epochs=10,
                    validation_data=(test_images, test_labels))
```

```python
# Graficamos la evolución de la accuracy y la loss
plt.figure(figsize=(12,5))

# Accuracy
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Entrenamiento')
plt.plot(history.history['val_accuracy'], label='Validación')
plt.title('Evolución de la Accuracy')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

# Loss
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Entrenamiento')
plt.plot(history.history['val_loss'], label='Validación')
plt.title('Evolución de la Loss')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.show()
```
