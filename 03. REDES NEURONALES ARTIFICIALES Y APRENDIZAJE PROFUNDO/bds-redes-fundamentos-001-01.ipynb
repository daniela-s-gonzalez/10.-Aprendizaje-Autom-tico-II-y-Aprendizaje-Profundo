 # **<font color="DeepPink">Fundamentos de Redes Neuronales y Deep Learning I</font>**

<p align="justify">
Una red neuronal es un modelo de computación inspirado en la estructura y el funcionamiento del cerebro humano. Está compuesta por unidades de procesamiento llamadas <B>neuronas</B> (<i>units</i>, <i>neurons</i>, <i>perceptrons</i> o <i>nodes</i>), organizadas en <b>capas</b> (<i>layers</i>). Cada neurona, realiza una operación sencilla y está conectada a las neuronas de la capa anterior y de la capa siguiente mediante <b>pesos</b> (<i>weights</i>), cuya función es regular la información que se propaga de una neurona a otra.
<br><br>
El Deep Learning (Aprendizaje Profundo) es un subcampo del Machine Learning que se basa en redes neuronales artificiales con muchas capas (<b>profundas</b>). Durante los últimos años, el interés y la aplicación de este tipo de modelos han experimentado tal expansión que se ha convertido en una disciplina por sí misma.
<BR><BR>
En este contexto, una neurona es una unidad de procesamiento que toma varias <b>entradas</b> (<i>inputs</i>), las combina a través de una función de <b>suma ponderada</b>, y luego aplica una <b>función de activación</b> para producir una <b>salida</b> (<i>output</i>). Estas neuronas son análogas a las neuronas biológicas, pero funcionan mediante operaciones matemáticas.

<p align="justify">
Las partes de una neurona son:
<ul p align="justify">
<li><b>Dentrita</b>: son extensiones ramificadas del soma. Reciben señales eléctricas de otras neuronas y las transmiten hacia el soma.
<li><b>Soma</b>: es el cuerpo celular de la neurona, es responsable de la síntesis de proteínas y otros componentes necesarios para su funcionamiento.
<li><b>Axon</b>: es una prolongación larga y delgada que se extiende desde el soma.Transmite señales eléctricas (potenciales de acción) desde el soma hacia otras neuronas, músculos o glándulas.





<p align="center">

<p align="justify">
En el ámbito de las redes neuronales artificiales, los términos, <b>dendrita</b>, <b>soma</b> y <b>axon</b> tienen analogías funcionales:
<ul p align="justify">
<li><b>Dentrita</b>: representan las conexiones de entrada a una neurona artificial. Reciben los valores de las neuronas de la capa anterior o de datos de entrada inicialmente proporcionados al sistema. Cada conexión tiene un <b>peso</b> (<i>weight</i>) asociado que determina la importancia de esa entrada en el proceso de suma ponderada.
<li><b>Soma</b>: corresponde al cuerpo de la neurona artificial. Realiza la suma ponderada de sus entradas (función de transferencia) y aplicación de la función de activación. La función de activación introduce no linealidades, permitiendo a la red aprender y modelar comportamientos complejos.
<li><b>Axon</b>: representa la salida de la neurona artificial. Transmite el resultado de la función de activación a las neuronas de la siguiente capa en la red.




## **<font color="DeepPink">Componentes de una neurona de la Red Neuronal Artificial</font>**

<ul p align="justify">
<li><b>Entradas (Inputs)</b>: valores numéricos provenientes de datos de entrada o de las salidas de otras neuronas. $x_1$, $x_2$,...,$x_n$
<li><b>Pesos (Weights)</b>: coeficientes que determinan la importancia de cada entrada. $w_1$, $w_2$,...,$w_n$
<li><b>Sesgo (Bias)</b>: un valor adicional que ajusta la salida de la neurona. $b_1$, $b_2$,...,$b_n$
<li><b>Suma ponderada</b>: función de transferencia, se multiplica cada valor de entrada $x_i$ por su peso asociado $w_i$ y se suman junto con el sesgo ($b$). $z= \sum^n_{i=1} x_i w_i + b$
<li><b>Función de activación</b>: una función no lineal que se aplica a la suma ponderada para introducir no linealidades en el modelo. Comúnmente utilizadas son ReLU o Sigmoid. $f(z)$
<li><b>Salidas (Outputs)</b>: serán las entradas para otras neuronas o predicción final del modelo.

## **<font color="DeepPink">Funcionamiento de una Red Neuronal Artificial (Forward Propagation)</font>**


<ol p align="justify">
<li><b>Recibir Entradas</b>: la neurona recibe múltiples entradas, cada una multiplicada por su peso correspondiente.
<li><b>Suma Ponderada</b>: Se calcula la suma de estos valores ponderados y se le añade el sesgo.
<li><b>Aplicar Función de Activación</b>: Se aplica una función de activación a la suma ponderada para obtener la salida final de la neurona.
<li><b>Producción de Salida</b>: La salida se pasa a las neuronas de la siguiente capa o se utiliza como salida final de la red (predicción).

<p align="center">
<img src="https://github.com/cristiandarioortegayubro/BDS/blob/main/images/RedNeuronal_01.png?raw=true" alt="Neural Network Example" width="500px">

<p align="justify">
La primera capa de la red neuronal se conoce como <b>capa de entrada</b> o <i>input layer</i> y recibe los datos en bruto, es decir, el valor de las variables predictoras. La capa intermedia, conocida como <b>capa oculta</b> o <i>hidden layer</i>, recibe los valores de entrada, ponderados por los pesos. La última capa, llamada <b>capa de salida</b> o <i>output layer</i>, combina los valores que salen de la última capa oculta para generar la predicción. En un problema de predicción la capa de salida contiene una sola neurona, mientras que en un problema de clasificacción contiene tantas neuronas como clases posea.

## **<font color="DeepPink">Proceso de entrenamiento de una Red Neuronal Artificial</font>**

<ol p align="justify">
<li><b>Inicialización</b>: los pesos y sesgos de las neuronas se inicializan aleatoriamente.
<li><b>Propagación hacia Adelante (Forward Propagation)</b>: los datos de entrada se pasan a través de la red para obtener la salida (predicción).
<li><b>Cálculo de Pérdida (Loss Calculation)</b>: se calcula la diferencia entre la salida de la red (pedicción) y la etiqueta real usando una <b>función de pérdida</b> (<i>lost function</i>).
<li><b>Propagación hacia Atrás (Backward Propagation)</b>: se calcula el gradiente de la función de pérdida con respecto a los pesos y sesgos, y se actualizan estos parámetros usando un algoritmo de optimización (como el <b>descenso del gradiente</b>).
<li><b>Iteración</b>: los pasos anteriores se repiten para múltiples iteraciones (<i>epochs</i>) hasta que la red neuronal se entrena adecuadamente.

## **<font color="DeepPink">Forward Propagation</font>**

<p align="justify">
En este notebook, se contruye una red neuronal desde cero y se codifican las predicciones mediante <b>Forward Propagation</b>. Tenga en cuenta que todas las bibliotecas de Redes Neuronales (Scikit-learn) y Deep Learning (Keras o Tensorflow) tienen implementados todos los procesos de entrenamiento y predicción, por lo que en la práctica no sería necesario construir una red neuronal desde cero. Sin embargo,  esta práctica ayuda a comprender aún mejor las redes neuronales y cómo funcionan.

<p align="justify">
A continuación se muetra cómo una <b>red neuronal</b> hace predicciones a través del proceso de <b>Forward Propagation</b>. Como ejemplo construimos una red neuronal que toma dos entradas, tiene una capa oculta (<i>hidden layer</i>) con dos neuronas y una capa de salida (<i>output layer</i>) con una neurona.

<p align="center">
<img src="http://cocl.us/neural_network_example" alt="Neural Network Example" width="500px">

<p align="justify">
Se comienza inicializando aleatoriamente los pesos (<i>weights</i>) y los sesgos (<i>biases</i>) en la red. La red tiene 6 pesos y 3 sesgos, uno para cada neurona de la capa oculta así como para la neurona de la capa de salida.

```python
import numpy as np                                                              # importamos numpy

np.random.seed(42)                                                              # inicializamos el generador de números aleatorios

weights = np.random.uniform(size=6).round(2)                                    # inicializamos los pesos
biases = np.random.uniform(size=3).round(2)                                     # inicializamos los sesgos

print(weights)
print(biases)
```

<p align="justify">
Ahora con los pesos y los sesgos definidos para la red, se calcula la salida para 2 entradas (inputs) dadas: $x_1$ y $x_2$.

```python
x_1 = 0.5   # input 1
x_2 = 0.85  # input 2

print('x1 es {} y x2 es {}'.format(x_1, x_2))
```

Se calcula la suma ponderada de las entradas en la primer neurona ($z_{1, 1}$) de la capa oculta.

```python
z_11 = x_1 * weights[0] + x_2 * weights[1] + biases[0]

print('La suma ponderada de las entradas en la primer neurona de la capa oculta es {}'.format(z_11))
```

A continuación, se calcula la suma ponderada de las entradas en la segunda neurona ($z_{1, 2}$) de la capa oculta.


```python
z_12 = x_1 * weights[2] + x_2 * weights[3] + biases[1]

print('La suma ponderada de las entradas en la segunda neurona de la capa oculta es {}'.format(z_12))
```

Luego, se asume una función de activación Sigmoid y se calcula la activación de la primer neurona de la capa oculta ($a_{1, 1}$).
<br><br>
Función Sigmoid (Sigmoide):

$$
a = f(z)=\frac{1}{1 + e^{-z}}
$$

Donde $e$ es el número de Euler, aproximadamente igual a 2.71828 y es la base del logaritmo natural.


```python
a_11 = 1.0 / (1.0 + np.exp(-z_11))

print('La activación de la primer neurona en la capa oculta es {}'.format(np.around(a_11, decimals=4)))
```

Se computa también la activación de la segunda neurona en la capa oculta ($a_{1, 2}$)


```python
a_12 = 1.0 / (1.0 + np.exp(-z_12))

print('La activación de la segunda neurona en la capa oculta es {}'.format(np.around(a_12, decimals=4)))
```

Ahora estas activaciones ($a_{1, 1}$ y $a_{1, 2}$) servirán como entradas (<i>inputs</i>) para la capa de salida (<i>output layer</i>). Se calcula la suma ponderada de esas entradas a la neurona de la capa de salida ($z_2$).



```python
z_2 = a_11 * weights[4] + a_12 * weights[5] + biases[2]

print('La suma ponderada de las entradas en la capa de salida es {}'.format(np.around(z_2, decimals=4)))
```

Finalmente se calcula la salida de la red como la activación del nodo en la capa de salida ($a_2 = f(z_2)$).

```python
a_2 = 1.0 / (1.0 + np.exp(-z_2))

print('La salida de la red para x1 = 0.5 y x2 = 0.85 es {}'.format(np.around(a_2, decimals=4)))
```

## **<font color="DeepPink">Forward Propagation Multicapa</font>**

<p align="justify">
Obviamente, las redes neuronales para problemas reales se componen de muchas capas ocultas y muchos más nodos en cada capa. Por lo tanto, no es posible seguir haciendo predicciones utilizando este enfoque tan ineficiente de calcular la suma ponderada en cada neurona y la activación de cada neurona manualmente.
<br><br>
Para codificar una forma automática de hacer predicciones, se generalizará la red. Una red general tomará $n$ entradas, tendrá muchas capas ocultas, cada capa oculta tendrá $m$ neuronas y tendrá una capa de salida. Aunque la red muestra una capa oculta, se codificará la red para que tenga muchas capas ocultas. De manera similar, aunque la red muestra una capa de salida con una neurona, se codificará la red para que tenga más de una neurona en la capa de salida.

<p align="center">
<img src="http://cocl.us/general_neural_network" alt="Neural Network General" width="500px">

<p align="justify">
Entonces, con la siguiente función, se pueden inicializar los pesos y los sesgos pertenecientes a cualquier red de cualquier número de capas ocultas y número de neuronas en cada capa.

```python
def initialize_network(num_inputs, num_hidden_layers, num_nodes_hidden, num_nodes_output):
    np.random.seed(42)
    num_nodes_previous = num_inputs                                             # número de nodos en la capa anterior
    network = {}                                                                # inicializamos la red como un diccionario vacio

    # recorrer cada capa e inicializar aleatoriamente los pesos y sesgos asociados con cada capa
    for layer in range(num_hidden_layers + 1):

        # determinamos el nombre de la capa
        if layer == num_hidden_layers:
            layer_name = 'output' # nombre de la última capa de la red
            num_nodes = num_nodes_output
        else:
            layer_name = 'layer_{}'.format(layer + 1) # de lo contrario, asigne un número a la capa
            num_nodes = num_nodes_hidden[layer]

        # inicializamos los pesos y sesgos para cada neurona
        network[layer_name] = {}
        for node in range(num_nodes):
            node_name = 'node_{}'.format(node+1)
            network[layer_name][node_name] = {
                'weights': np.around(np.random.uniform(size=num_nodes_previous), decimals=2),
                'bias': np.around(np.random.uniform(size=1), decimals=2),
            }

        num_nodes_previous = num_nodes

    return network # devuelve la red
```

<p align="justify">
A continuación se define formalmente la estructura de la red (arquitectura). Se usa la función <code>initialize_network</code> para crear una red que toma 5 entradas, tiene 3 capas ocultas. Además tiene 3 neuronas en la primer capa (oculta), 2 en la segunda capa y 3 en la tercera. Tiene una sola neurona en la capa de salida (problema de regresión).

```python
import pprint
pp = pprint.PrettyPrinter(indent=4)

small_network = initialize_network(5, 3, [3, 2, 3], 1)
pp.pprint(small_network)
```

Se generan 5 entradas:

```python
from random import seed
import numpy as np

np.random.seed(12)
inputs = np.around(np.random.uniform(size=5), decimals=2)

print('Las entradas de la red son {}'.format(inputs))
```


La suma ponderada en cada nodo se calcula como el producto escalar de las entradas y las ponderaciones más el sesgo. Entonces, se crea una función llamada `compute_weighted_sum` que hace precisamente eso.

```python
def compute_weighted_sum(inputs, weights, bias):
    return np.sum(inputs * weights) + bias
```

Se utiliza la función `compute_weighted_sum` para calcular la suma ponderada en la primer neurona de la primer capa oculta.

```python
node_weights = small_network['layer_1']['node_1']['weights']
node_bias = small_network['layer_1']['node_1']['bias']

weighted_sum = compute_weighted_sum(inputs, node_weights, node_bias)
print('La suma ponderada del primer nodo en la primer capa oculta es {}'.format(np.around(weighted_sum[0], decimals=4)))
```

<p align="justify">
Recuerde que la salida de cada neurona es simplemente una transformación no lineal de la suma ponderada. Se utilizan las funciones de activación para este mapeo. Se usará la función Sigmoid. Entonces, se define una función que toma una suma ponderada como entrada y devuelve la transformación no lineal de la entrada usando la función Sigmoid.

```python
def node_activation(weighted_sum):
    return 1.0 / (1.0 + np.exp(-1 * weighted_sum))
```

Se utiliza la función de activación Sigmoid para calcular la salida de la primer neurona en la primera capa oculta.

```python
node_output  = node_activation(compute_weighted_sum(inputs, node_weights, node_bias))
print('La salida de la primer neurona de la primer capa oculta es {}'.format(np.around(node_output[0], decimals=4)))
```

<p align="justify">
Por último, para la construcción de una red neuronal que pueda realizar predicciones se debe juntar todo. Entonces, se creará una función que aplique las funciones <code>compute_weighted_sum</code> y <code>node_activation</code> a cada neurona de la red y propague los datos hasta la capa de salida y genere una predicción para cada neurona en la capa de salida.

La forma en que se logra esto es mediante el siguiente procedimiento:

1. Se comienza con la <b>capa de entrada</b> como entrada a la primer <b>capa oculta</b>.
2. Se calcula la <b>suma ponderada</b> en las neuronas de la capa actual.
3. Se calcula la salida de las neuronas de la capa actual mediante una función de activación.
4. Las salidas de la capa actual son la entradas de la siguiente capa. Se pasa a la siguiente capa de la red.
5. Se repiten los pasos 2 a 4 hasta que se determina la salida de la <b>capa de salida</b> (predicción).

```python
def forward_propagate(network, inputs):

    layer_inputs = list(inputs) # comience con la capa de entrada como entrada a la primera capa oculta

    for layer in network:

        layer_data = network[layer]

        layer_outputs = []
        for layer_node in layer_data:

            node_data = layer_data[layer_node]

            # calcular la suma ponderada y la salida de cada nodo al mismo tiempo
            node_output = node_activation(compute_weighted_sum(layer_inputs, node_data['weights'], node_data['bias']))
            layer_outputs.append(np.around(node_output[0], decimals=4))

        if layer != 'output':
            print('Las salidas de los nodos en la capa oculta número {} son {}'.format(layer.split('_')[1], layer_outputs))

        layer_inputs = layer_outputs # establecer la salida de esta capa para que sea la entrada a la siguiente capa

    network_predictions = layer_outputs
    return network_predictions
```

Se utiliza la función `forward_propagate` para calcular la predicción de esta pequeña red.

```python
predictions = forward_propagate(small_network, inputs)
print('El valor predicho por la red para la entrada dada es {}'.format(np.around(predictions[0], decimals=4)))
```
