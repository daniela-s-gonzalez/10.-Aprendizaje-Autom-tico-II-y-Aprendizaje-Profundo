 # **<font color="DeepPink">Deep Learning con Keras</font>**

<p align="justify">
Keras es una biblioteca para Aprendizaje Profundo (Deep Learning) que proporciona una interfaz de alto nivel para construir y entrenar modelos de redes neuronales. Es fácil de usar, modular y extensible, lo que la convierte en una opción popular tanto para investigadores como para desarrolladores.
<br><br>
Keras está escrita en Python y es capaz de ejecutarse sobre otras bibliotecas de bajo nivel como TensorFlow o Theano, permitiendo a los usuarios aprovechar las capacidades avanzadas de estas plataformas sin tener que lidiar con su complejidad.
<br><br>
Algunas características clave de Keras incluyen:
<ul p align="justify">
<li><b>Simplicidad y facilidad de uso</b>: está diseñada para permitir una rápida experimentación. Su interfaz intuitiva permite crear y entrenar modelos de redes neuronales con pocas líneas de código.
<li><b>Modularidad</b>: es altamente modular, lo que permite combinar diferentes bloques de construcción como capas, funciones de activación, y algoritmos de optimización para crear modelos personalizados.
<li><b>Extensibilidad</b>: los usuarios pueden extender Keras para implementar nuevas capas, funciones de pérdida, métricas y otros componentes.
<li><b>Compatibilidad</b>: es compatible con múltiples backend de Deep Learning, lo que proporciona flexibilidad y la capacidad de aprovechar las optimizaciones específicas del backend subyacente.
</ul>

<p align="justify">
En resumen, Keras facilita el desarrollo y la experimentación con modelos de Deep Learning, haciendo que la construcción de redes neuronales complejas sea accesible y eficiente.

 ## **<font color="DeepPink">Problema de Regresión con Keras</font>**

<p align="justify">
En este ejemplo, se mostrará cómo usar la biblioteca Keras para construir un modelo de regresión.


```python
import pandas as pd
import numpy as np

import tensorflow as tf
# Establecer semillas globales
np.random.seed(42)
tf.random.set_seed(42)
```

```python
data = pd.read_csv('https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/DL0101EN/labs/data/concrete_data.csv')
data.head()
```

<p align="justify">
El conjunto de datos está relacionado con la composición y las propiedades del concreto. Específicamente, se trata de un conjunto de datos que contiene información sobre las proporciones de diferentes materiales utilizados en la mezcla de concreto y la resistencia a la compresión del concreto resultante después de un cierto período de tiempo. Aquí hay un desglose de variables y su significado:
<ul p align="justify">
<li><code>Cement</code>: cantidad de cemento en metros cúbicos.
<li><code>Blast Furnace Slag</code>: cantidad de escoria de alto horno en metros cúbicos.
<li><code>Fly Ash</code>: cantidad de cenizas volantes en metros cúbicos.
<li><code>Water</code>: cantidad de agua en metros cúbicos.
<li><code>Superplasticizer</code>: cantidad de superplastificante en metros cúbicos.
<li><code>Coarse Aggregate</code>: cantidad de agregado grueso en metros cúbicos.
<li><code>Fine Aggregate</code>: cantidad de agregado fino en metros cúbicos.
<li><code>Age</code>: edad del concreto.
<li><code>Strength</code>: La resistencia a la compresión del concreto en megapascales (MPa).
</ul>
<p align="justify">
<br>
Este tipo de conjunto de datos se utiliza comúnmente en la ingeniería civil y la investigación de materiales para analizar y predecir las propiedades del concreto basándose en sus componentes y condiciones de curado. Los modelos de Machine Learning pueden ser aplicados a estos datos para predecir la resistencia a la compresión del concreto a partir de las proporciones de los materiales utilizados en la mezcla.
<br><br>
Entonces, la primera muestra concreta tiene 540 metros cúbicos de cemento, 0 metros cúbicos de escoria de alto horno, 0 metros cúbicos de cenizas volantes, 162 metros cúbicos de agua, 2.5 metros cúbicos de superplastificante, 1040 metros cúbicos de agregado grueso y 676 metros cúbicos de agregado fino. Una mezcla de concreto de este tipo, con 28 días de antigüedad, tiene una resistencia a la compresión de 79.99 MPa.

```python
data.shape
```

<p align="justify">
Hay aproximadamente 1000 muestras para entrenar el modelo. Debido al número reducido de muestras, se debe tener cuidado de no sobreajustar los datos de entrenamiento.

Se analiza los estadísticos del DataFrame y si existen valores faltantes.


```python
data.describe().T
```

```python
data.isnull().sum()
```

<p align="justify">
La variable objetivo en este problema es la resistencia de la muestra de concreto (<code>Strength</code>). Por lo tanto, las variables predictoras serán todas las demás columnas.








```python
X = data.drop(columns='Strength')                                               # todas las columnas excepto Strength
y = data['Strength']                                                            # columna Strength
```

```python
X.head()
```

```python
y.head()
```

<p align="justify">
Se normalizan los datos restando la media y dividiendo por la desviación estándar.

```python
X_norm = (X - X.mean()) / X.std()
X_norm.head()
```

```python
X_norm.describe().T.round(2)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_norm, y, test_size=0.25, random_state=42)
```

<p align="justify">
Se guarda el número de  variables predictoras en <code>n_cols</code> ya que  este número será necesario al construir la red neuronal.

```python
n_cols = X_train.shape[1]                                                       # Número de variables predictoras
n_cols
```

```python
import keras
```

```python
print(keras.__version__)
```

```python
# Inicializadores con semillas para asegurar que la inicialización de los pesos sea reproducible
from keras.initializers import GlorotUniform
```

Se importan el resto de los paquetes de la biblioteca Keras necesarios para construir nuestro modelo de regresión.

```python
from keras.models import Sequential
from keras.layers import Dense
```

<p align="justify">
<code>Secuential</code> se utiliza para crear modelos de redes neuronales secuenciales en Keras. Un modelo secuencial es simplemente una pila lineal de capas donde la salida de una capa se convierte en la entrada de la siguiente capa. Es adecuado para la mayoría de los modelos simples y es fácil de usar.
<BR><BR>
La clase <code>Dense</code> se utiliza para crear capas densas (totalmente conectadas) en la red neuronal. Una capa Dense es una capa en la que cada neurona está conectada a todas las neuronas de la capa anterior. Es una de las capas más comunes en las redes neuronales.

Se contruye una red neuronal con la siguiente arquitectura:


<p align="center">
<img src="https://github.com/cristiandarioortegayubro/BDS/blob/main/images/ejemploRed1.jpg?raw=true" width="500px">
</p>


```python
model_r = Sequential()
model_r.add(Dense(5, activation='relu', kernel_initializer=GlorotUniform(seed=42), input_shape=(n_cols,))) # Se añade una capa con 5 neuronas.
model_r.add(Dense(5, activation='relu', kernel_initializer=GlorotUniform(seed=42))) # Se añade una capa con 5 neuronas adicionales.
model_r.add(Dense(1, kernel_initializer=GlorotUniform(seed=42)))                # Se añade una capa con 1 neurona. Dado que no se especifica una función de activación, se usará la función de activación lineal por defecto, lo cual es adecuado para problemas de regresión.
model_r.compile(optimizer='adam', loss='mean_absolute_error')                   # se complila el modelo.
```

<p align="justify">
El método <CODE>add()</CODE> para agregar cada capa densa. Se especifica el número de neuronas en cada capa y la función de activación a utilizar. La función ReLU es una de las funciones de activación recomendadas para capas ocultas. Y para la primera capa oculta es necesario pasar el parámetro <code>input_shape</code>, que es el número de columnas o variables predictoras en este conjunto de datos. Luego se repite lo mismo para la otra capa oculta, sin el parámetro <code>input_shape</code>, y por último, se crea la capa de salida con un nodo.
<br><br>
Para (<code>compile</code>) se debe definir un optimizador y la métrica de error. En notebooks anteriores, se utilizó el descenso del gradiente como algoritmo. Existen algoritmos más eficientes que el descenso de gradiente para Deep Learning. Uno de ellos es 'adam'. Una de las principales ventajas del optimizador 'adam' es que no es necesario especificar la tasa de aprendizaje (learning rate) utilizada en el descenso del gradiente.


<p align="justify">
Por último, se entrana y testea el modelo
<BR><BR>
Se utiliza el ajuste (<CODE>fit</CODE>) para entrenar nuestro modelo. Una vez entrenando, se pueden hacer predicciones utilizando el método <CODE>predict</CODE>.


```python
model_r.fit(X_train, y_train, validation_split=0.3, epochs=10)                  # se entrena el modelo con 10 iteraciones.
```

```python
model_r.evaluate(X_test, y_test)                                                # se evalua el modelo con los datos de prueba
```

```python
model_r.predict(X_test)[:10]                                                    # se predice el modelo con los datos de prueba
```

```python
tabla = pd.DataFrame({'y_test': y_test,
              'y_pred': model_r.predict(X_test).reshape(-1),
              'error': abs(model_r.predict(X_test).reshape(-1) - y_test)})
tabla.head()
```

```python
tabla.error.mean().round(2)
```

Se agrega una capa oculta más y 50 neuronas por capa, además se realizan 100 epochs.

```python
model = Sequential()
model.add(Dense(50, activation='relu', kernel_initializer=GlorotUniform(seed=42), input_shape=(n_cols,)))
model.add(Dense(50, activation='relu', kernel_initializer=GlorotUniform(seed=42)))
model.add(Dense(50, activation='relu', kernel_initializer=GlorotUniform(seed=42)))
model.add(Dense(1, kernel_initializer=GlorotUniform(seed=42)))
model.compile(optimizer='adam', loss='mean_absolute_error')
model.fit(X_train, y_train, validation_split=0.3, epochs=100)
```

```python
model.evaluate(X_test, y_test)
```

```python
model.predict(X_test)[:10]
```
